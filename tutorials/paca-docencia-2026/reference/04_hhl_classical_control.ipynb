Se resuelve Ax=b clásicamente como referencia del tutorial HHL histórico. No se implementa HHL ni se presume ventaja cuántica. El tutorial HHL existente conserva su atribución a Qiskit Textbook.


```python
import numpy as np
A=np.array([[1.,-1/3],[-1/3,1.]])
b=np.array([1.,0.])
x=np.linalg.solve(A,b)
residual=np.linalg.norm(A@x-b)
kappa=np.linalg.cond(A)
assert np.allclose(x,np.array([1.125,0.375]))
assert residual<1e-12 and kappa>=1
print('x:',x,'cond:',round(kappa,4))
```
