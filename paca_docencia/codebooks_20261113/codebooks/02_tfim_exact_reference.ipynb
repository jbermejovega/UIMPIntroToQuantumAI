Hamiltoniano abierto H = -J sum_i Z_i Z_(i+1) - h sum_i X_i. Diagonalización numérica para n <= 5 como referencia para VQE y DMRG; no se han ejecutado estos algoritmos aquí.


```python
import numpy as np
I=np.eye(2)
X=np.array([[0.,1.],[1.,0.]])
Z=np.diag([1.,-1.])
def operator_at(paulis):
    result=np.array([[1.]])
    for p in paulis:
        result=np.kron(result,p)
    return result
def tfim_hamiltonian(n,J=1.,h=0.5):
    if not 2<=n<=5:
        raise ValueError('Se admiten 2..5 qubits')
    H=np.zeros((2**n,2**n))
    for i in range(n-1):
        H-=J*operator_at([Z if j in (i,i+1) else I for j in range(n)])
    for i in range(n):
        H-=h*operator_at([X if j==i else I for j in range(n)])
    return H
n=4
H=tfim_hamiltonian(n)
eigvals=np.linalg.eigvalsh(H)
assert np.allclose(H,H.T)
assert np.isclose(np.linalg.eigvalsh(tfim_hamiltonian(n,J=1,h=0))[0],-(n-1))
assert np.isclose(np.linalg.eigvalsh(tfim_hamiltonian(n,J=0,h=1))[0],-n)
print('E0:',round(eigvals[0],8))
```
