Un estado puro de tres qubits se reorganiza en matriz 2 x 4. Su SVD proporciona los coeficientes de Schmidt para 1|23. El error de truncar a rango uno coincide con la norma de los singulares descartados. No es un DMRG completo.


```python
import numpy as np
psi=np.array([1.,0.,0.,0.,0.,0.,0.,1.],dtype=complex)
psi/=np.linalg.norm(psi)
A=psi.reshape(2,4)
U,s,Vh=np.linalg.svd(A,full_matrices=False)
A_full=(U*s)@Vh
A_rank1=np.outer(U[:,0]*s[0],Vh[0,:])
err=np.linalg.norm(A-A_rank1)
assert np.allclose(A,A_full)
assert np.isclose(err,np.linalg.norm(s[1:]))
assert np.isclose(np.sum(s**2),1)
print('Schmidt:',s,'error rango 1:',err)
```
